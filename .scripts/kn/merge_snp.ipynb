```python
#!/usr/bin/env python3
"""
Merge all sutta files inside each vagga folder of Kinh Tiểu Bộ (Snp)
into one .md file per vagga.

Input : docs/kinhtieubo/pali-vi/snp/vagga{N}/snp-*.md
Output: docs/kinhtieubo/pali-vi/snp/vagga{N}-{slug}.md
"""

import re
from pathlib import Path

BASE = Path("/Users/ng/projects/nikaya2/docs/kinhtieubo/pali/iti")
OUT  = Path("/Users/ng/projects/nikaya2/docs/kinhtieubo/pali-vi/iti")
# idx -> (Pali display name, slug for filename)
VAGGAS = {
    1: ("Uragavagga",    "uragavagga"),
    2: ("Cūḷavagga",     "culavagga"),
    3: ("Mahāvagga",     "mahavagga"),
    4: ("Aṭṭhakavagga",  "atthakavagga"),
    5: ("Pārāyanavagga", "parayanavagga"),
}

VAGGAS = {
    1: ("Bodhivagga",          "bodhivagga"),
    2: ("Mucalindavagga",      "mucalindavagga"),
    3: ("Nandavagga",          "nandavagga"),
    4: ("Meghiyavagga",        "meghiyavagga"),
    5: ("Soṇavagga",           "sonavagga"),
    6: ("Jaccandhavagga",      "jaccandhavagga"),
    7: ("Cūḷavagga",           "culavagga"),
    8: ("Pāṭaligāmiyavagga",   "pataligamiyavagga"),
}

VAGGAS = {
    1:  ("Paṭhamavagga",              "ekakanipata-pathamavagga"),
    2:  ("Dutiyavagga",               "ekakanipata-dutiyavagga"),
    3:  ("Tatiyavagga",               "ekakanipata-tatiyavagga"),
    4:  ("Paṭhamavagga",              "dukanipata-pathamavagga"),
    5:  ("Dutiyavagga",               "dukanipata-dutiyavagga"),
    6:  ("Paṭhamavagga",              "tikanipata-pathamavagga"),
    7:  ("Dutiyavagga",               "tikanipata-dutiyavagga"),
    8:  ("Tatiyavagga",               "tikanipata-tatiyavagga"),
    9:  ("Catutthavagga",             "catutthavagga"),
    10: ("Pañcamavagga",              "pancamavagga"),
    11: ("Brāhmaṇadhammayāgavagga",   "brahmanadhammayagavagga"),
}




def natural_key(path: Path):
    """Sort snp-1-2 < snp-1-10 (numeric aware)."""
    return [int(t) if t.isdigit() else t.lower()
            for t in re.split(r"(\d+)", path.name)]


def promote_heading(text: str) -> str:
    """Turn the first '# SNP x.y. ...' line into '## SNP x.y. ...'."""
    lines = text.split("\n")
    for i, line in enumerate(lines):
        if line.startswith("# "):
            lines[i] = "#" + line          # "# " -> "## "
            break
        if line.strip():                    # non-empty non-heading line => stop
            break
    return "\n".join(lines)


def merge_vagga(idx: int, pali_name: str, slug: str) -> None:
    folder = BASE / f"vagga{idx}"
    if not folder.is_dir():
        print(f"⚠️  Skip: {folder} not found")
        return

    files = sorted(folder.glob("iti-*.md"), key=natural_key)
    if not files:
        print(f"⚠️  Skip: no .md files in {folder}")
        return

    chunks = [f"# {idx}. {pali_name}"]
    for f in files:
        content = f.read_text(encoding="utf-8").rstrip()
        chunks.append(promote_heading(content))

    out_path = OUT / f"vagga{idx}-{slug}.md"
    out_path.write_text("\n\n".join(chunks) + "\n", encoding="utf-8")
    print(f"✓ {out_path}  ← {len(files)} file(s)")



for i, (pali, slug) in VAGGAS.items():
    merge_vagga(i, pali, slug)
```
