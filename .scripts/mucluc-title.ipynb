```python
from pathlib import Path

# Thư mục gốc cần quét
ROOT = Path("../docs")   # đổi thành Path("/đường/dẫn") nếu cần

HEADER = "---\ntitle: Mục lục\n---\n\n"

updated = []
skipped = []
errors = []

for file in ROOT.rglob("mucluc.md"):
    try:
        text = file.read_text(encoding="utf-8")

        # File đã có frontmatter
        if text.startswith("---"):
            skipped.append(file)
            print(f"[SKIP] {file}")
            continue

        file.write_text(HEADER + text, encoding="utf-8")
        updated.append(file)
        print(f"[UPDATE] {file}")

    except Exception as e:
        errors.append((file, str(e)))
        print(f"[ERROR] {file}: {e}")

print("\n========== SUMMARY ==========")
print(f"Updated : {len(updated)}")
print(f"Skipped : {len(skipped)}")
print(f"Errors  : {len(errors)}")

if skipped:
    print("\nFiles already containing frontmatter:")
    for f in skipped:
        print(" -", f)

if errors:
    print("\nErrors:")
    for f, e in errors:
        print(f" - {f}: {e}")
```

```python
# thay header kinh tieu bo
# Cell 1: Import thư viện và cấu hình
import os
import re
from pathlib import Path

# ===== CẤU HÌNH =====
FILES = [

# "/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-054-tap-9-pham-1-tap-mot-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-055-tap-9-pham-2-tap-hai-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-056-tap-9-pham-3-tap-ba-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-057-tap-9-pham-4-tap-bon-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-058-tap-9-pham-5-tap-nam-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-059-tap-9-pham-6-tap-sau-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-060-tap-9-pham-7-tap-bay-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-061-tap-9-pham-8-tap-tam-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-062-tap-9-pham-9-tap-chin-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-063-tap-9-pham-10-tap-muoi-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-064-tap-9-pham-11-tap-muoi-hai-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-065-tap-9-pham-12-tap-muoi-sau-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-066-tap-9-pham-13-tap-hai-muoi-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-067-tap-9-pham-14-tap-ba-muoi-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-068-tap-9-pham-15-tap-bon-muoi-ke.md",
"/Users/ng/projects/nikaya2/docs/kinhtieubo/thichminhchau/kn-069-tap-9-pham-16-dai-pham.md",
]  # Danh sách files cần xử lý
# Prefix cho title
TITLE_PREFIX = "Trưởng Lão Ni Kệ - "

# Nav HTML block
NAV_BLOCK = """<div class="top-nav">

[Kinh tiểu bộ](/kinhtieubo/thichminhchau/) > [Trưởng Lão Ni Kệ](/kinhtieubo/thichminhchau/kn-053-tap-9-truong-lao-ni-ke)

</div>"""

# Cell 2: Hàm xử lý
def replace_header(file_path: str, title_prefix: str, nav_block: str, dry_run: bool = False):
    """
    Đọc file .md, thay H1 header đầu tiên bằng frontmatter + nav + H1.
    """
    p = Path(file_path)
    if not p.exists():
        print(f"❌ Không tìm thấy: {file_path}")
        return

    content = p.read_text(encoding="utf-8")

    # Tìm H1 đầu tiên (# ...)
    match = re.search(r"^#\s+(.+?)\s*$", content, flags=re.MULTILINE)
    if not match:
        print(f"⚠️  Không có H1 trong: {file_path}")
        return

    original_header = match.group(1).strip()
    new_title = f"{title_prefix}{original_header}"

    # Tạo block mới
    new_block = (
        f"---\n"
        f"title: {new_title}\n"
        f"---\n\n"
        f"{nav_block}\n\n"
        f"# {original_header}\n"
    )

    # Thay thế H1 đầu tiên bằng block mới
    new_content = content[:match.start()] + new_block + content[match.end():]

    if dry_run:
        print(f"--- DRY RUN: {file_path} ---")
        print(new_content[:400], "...\n")
    else:
        p.write_text(new_content, encoding="utf-8")
        print(f"✅ Đã xử lý: {file_path}")

    return new_content

# Cell 3: Chạy thử (dry run) để kiểm tra
for f in FILES:
    replace_header(f, TITLE_PREFIX, NAV_BLOCK, dry_run=False)
```
