```python
import os
import re

def generate_filelist(folder_path, listname, folder_out):
    """
    filelist.js using for generate next/back button config in config.js
    mucluc.md using for generate table of contents

    Generates a JavaScript filelist from Markdown files in a folder.

    Args:
        folder_path: The path to the folder containing the Markdown files.
        listname: The name to use for the JavaScript constant and export.

    Returns:
        None.  Writes the output to a file named 'filelist.js' in the input folder.
    """
    os.makedirs(os.path.join(folder_path, 'meta'), exist_ok=True)


    file_list = []
    _lst = sorted([f for f in os.listdir(folder_path) if f.endswith('.md')])
    for filename in _lst:
        # print(filename)
        file_path = os.path.join(folder_path, filename)
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                title = None

                for line in f:
                    # Match header Markdown: #, ##, ###, ...
                    match = re.match(r"^#+\s+(.*)", line.strip())
                    if match:
                        title = match.group(1).strip()
                        break

                if title:
                    file_item = {
                        "text": title,
                        "link": os.path.join(folder_out, filename).replace("\\", "/")
                    }
                    file_list.append(file_item)
                    print(file_item)
                else:
                    print(f"Warning: No title found in {filename}, see `9headerformat.ipynb`")

        except Exception as e:
            print(f"Error reading {filename}: {e}")

        except FileNotFoundError:
            print(f"Error: File not found: {file_path}")
        except Exception as e:
            print(f"Error reading or processing {filename}: {e}")

    # Sort the file list alphabetically by 'text'
    # file_list.sort(key=lambda x: x['text'])

    # Generate the JavaScript code.
    js_code = f"const {listname} = [\n"
    for item in file_list:
        js_code += f'  {{ text: "{item["text"]}", link: "{item["link"]}" }},\n'
    js_code += "];\n"
    js_code += f"export default {listname};\n"

    meta_path = folder_path+"/meta"
    # Write the JavaScript code to filelist.js
    output_file_path = os.path.join(meta_path, "filelist.js")
    try:
        with open(output_file_path, 'w', encoding='utf-8') as outfile:
            outfile.write(js_code)
        print(f"Successfully generated filelist.js in {meta_path}")
    except Exception as e:
        print(f"Error writing to filelist.js: {e}")

    # Generate the Markdown table of contents (mucluc.md)
    md_toc = ""
    for item in file_list:
        md_toc += f'- [{item["text"]}]({item["link"]})\n'

    # Write the Markdown TOC to mucluc.md
    output_md_path = os.path.join(meta_path, "mucluc.md")
    try:
        with open(output_md_path, 'w', encoding='utf-8') as outfile:
            outfile.write(md_toc)
        print(f"Successfully generated mucluc.md in {folder_path}")
    except Exception as e:
        print(f"Error writing to mucluc.md: {e}")

    # Generate index.md
    output_index_path = os.path.join(folder_path, "index.md")
    try:
        with open(output_index_path, 'w', encoding='utf-8') as outfile:
            outfile.write("<!--@include: ./meta/mucluc.md-->")
        print(f"Successfully generated index.md in {folder_path}")
    except Exception as e:
        print(f"Error writing to index.md: {e}")


    print(f"Created {output_file_path} for vite config.js create next-prev link")
    print(f"Created {output_md_path} for for mucluc.md ")




## step 3 - MAKE MUCLUC-NAVFILE


alang = ""
kinh = "kinhtieubo"
author="pali"
if alang != "":
    # folder contain list of file
    #folder_for_scan = f"../docs/{kinh}/{author}-{alang}/intro"
    folder_for_scan = f"../docs/{kinh}/{author}-{alang}"
    # output variable in filelist.js
    print_listname = f"{kinh}_{author}_{alang}"
    # output variable in filelist.js
    #print_folder_path = f'/{kinh}/{author}-{alang}/intro'
    print_folder_path = f'/{kinh}/{author}-{alang}'
else:
    folder_for_scan = f"../docs/{kinh}/{author}"
    # output variable in filelist.js
    print_listname = f"{kinh}_{author}"
    # output variable in filelist.js
    print_folder_path = f'/{kinh}/{author}'


folder_for_scan = f"../docs/the-life-of-the-buddha"
print_listname = f"life_of_budda"
print_folder_path = f'/the-life-of-the-buddha'

if not os.path.isdir(folder_for_scan):
    print(f"Error: The provided folder path '{folder_for_scan}' is not a valid directory.")
else:
    generate_filelist(folder_for_scan, print_listname, print_folder_path)
    print("Remember add mucluc.md to url-able path; and add filelist.js to config.ts for next-prev button")

# kinh_lst = [
#    'iti'
# ]

# for k in range(1, 9):
#   folder_for_scan = f"../docs/kinhtieubo/pali/ud/vagga{k}"
#   print_listname = f"kn_ud_vagga{k}"
#   print_folder_path = f'/kinhtieubo/pali/ud/vagga{k}'

#   if not os.path.isdir(folder_for_scan):
#       print(f"Error: The provided folder path '{folder_for_scan}' is not a valid directory.")
#   else:
#       generate_filelist(folder_for_scan, print_listname, print_folder_path)
#       print("Remember add mucluc.md to url-able path; and add filelist.js to config.ts for next-prev button")


```

```python
# đổi tên file 
import re
import sys
import unicodedata
from pathlib import Path


H1_RE = re.compile(r"^#\s+(.+?)\s*$", re.MULTILINE)
PREFIX_RE = re.compile(r"^(\d+(?:[-_]\d+)?)")


def slugify(text: str) -> str:
    # Bỏ markdown cơ bản
    text = re.sub(r"!\[.*?\]\(.*?\)", "", text)
    text = re.sub(r"\[([^\]]+)\]\([^)]+\)", r"\1", text)
    text = re.sub(r"[*_`~]", "", text)

    # Unicode -> ASCII
    text = unicodedata.normalize("NFKD", text)
    text = text.encode("ascii", "ignore").decode("ascii")

    text = text.lower()
    text = re.sub(r"[^a-z0-9]+", "-", text)

    return text.strip("-")


def get_prefix(filename: str) -> str:
    """
    Lấy 2 số đầu ở đầu filename.

    Ví dụ:
        01-05-abc.md  -> 01-05
        01_05-abc.md  -> 01_05
        12-abc.md      -> 12
    """
    match = PREFIX_RE.match(Path(filename).stem)

    if match:
        return match.group(1)

    return ""


def rename_file(file_path: Path):
    text = file_path.read_text(encoding="utf-8")

    # H1 đầu tiên
    match = H1_RE.search(text)

    if not match:
        print(f"[SKIP] Không tìm thấy H1: {file_path}")
        return

    h1 = match.group(1)
    slug = slugify(h1)

    if not slug:
        print(f"[SKIP] H1 không tạo được slug: {file_path}")
        return

    prefix = get_prefix(file_path.name)

    if prefix:
        new_name = f"{prefix}-{slug}.md"
    else:
        new_name = f"{slug}.md"

    new_path = file_path.with_name(new_name)

    if new_path == file_path:
        print(f"[OK]     {file_path.name}")
        return

    if new_path.exists():
        print(f"[ERROR]  Đã tồn tại: {new_path.name}")
        return

    file_path.rename(new_path)

    print(f"[RENAME] {file_path.name}")
    print(f"      -> {new_path.name}")



def smain():
    filenames =[
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/05-key-to-abbreviations.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/06-editor-s-preface.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/07-note-to-the-third-edition.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/08-introduction.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/09-voices.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/10-1-the-birth-and-the-early-years.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/11-2-the-struggle-for-enlightenment.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/12-3-after-the-enlightenment.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/13-4-the-spreading-of-the-dhamma.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/14-5-the-two-chief-disciples.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/15-6-anathapindika.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/16-7-formation-of-the-order-of-nuns.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/17-8-the-quarrel-at-kosambi.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/18-9-the-end-of-the-first-twenty-years.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/19-10-the-middle-period.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/20-11-the-person.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/21-12-the-doctrine.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/22-13-devadatta.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/23-14-old-age.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/24-15-the-last-year.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/25-16-the-first-council.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/26-list-of-sources.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/27-udana.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/28-index-items.md",
"/Users/ng/projects/nikaya2/docs/the-life-of-the-buddha/29-about-the-author.md",
        
    ]

    for filename in filenames:
        path = Path(filename)

        if not path.exists():
            print(f"[ERROR] Không tồn tại: {path}")
            continue

        if path.suffix.lower() != ".md":
            print(f"[SKIP] Không phải .md: {path}")
            continue

        rename_file(path)

smain()
```
