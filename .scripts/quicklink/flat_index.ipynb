```python
import json

def flatten_object(data):
    result = {}

    for parent in data.values():
        title = parent.get("title")
        slug = parent.get("slug")

        for key, child in parent.get("children", {}).items():
            anchor = child.get("anchor")

            result[key] = {
                "title": title,
                "slug": f"{slug}#{anchor}"
            }

    return result

items = flatten_object(data)

print(json.dumps(items, ensure_ascii=False, indent=2))
```
