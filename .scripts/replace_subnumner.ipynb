```python
from pathlib import Path
import re

FILE = Path("/Users/ng/projects/nikaya2/.docsource/giao-ly-va-truong-phai/giao-ly-va-truong-phai.md")

SUPERSCRIPT_MAP = str.maketrans({
    "⁰": "0",
    "¹": "1",
    "²": "2",
    "³": "3",
    "⁴": "4",
    "⁵": "5",
    "⁶": "6",
    "⁷": "7",
    "⁸": "8",
    "⁹": "9",
})

pattern = re.compile(r"[⁰¹²³⁴⁵⁶⁷⁸⁹]+")

text = FILE.read_text(encoding="utf-8")

def replace(match):
    number = match.group(0).translate(SUPERSCRIPT_MAP)
    return f"[^{number}]"

new_text, count = pattern.subn(replace, text)

FILE.write_text(new_text, encoding="utf-8")

print(f"Đã thay {count} lần trong {FILE}")
```
