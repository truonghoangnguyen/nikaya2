```python
from pathlib import Path
import re
import unicodedata


# =========================
# CONFIG
# =========================

input_file = Path("./out/output-all-refine.md")

# Số thứ tự bắt đầu
init_num = 5

# Header cấp nào dùng để chia file
# 1 = "# Header"
# 2 = "## Header"
header_level = 1

# True = tạo folder output riêng
create_output_folder = False

# Nếu True, file sẽ có dạng:
# 04-editor-preface.md
# Nếu False:
# 04.md
include_header_slug = True


# =========================
# FUNCTIONS
# =========================

def slugify(text: str) -> str:
    """
    Chuyển header thành filename slug.
    Ví dụ:
        EDITOR'S PREFACE
        -> editor-s-preface
    """
    text = text.strip()

    # Loại markdown formatting nếu có
    text = re.sub(r'[*_`~]', '', text)

    # Normalize Unicode
    text = unicodedata.normalize("NFKD", text)

    # Bỏ dấu
    text = text.encode("ascii", "ignore").decode("ascii")

    # Apostrophe -> khoảng trắng
    text = text.replace("'", " ")

    # Chỉ giữ chữ/số
    text = re.sub(r"[^a-zA-Z0-9]+", "-", text)

    # Xóa dấu - ở đầu/cuối
    text = text.strip("-")

    return text.lower()


def split_markdown(input_path: Path):
    content = input_path.read_text(encoding="utf-8")

    # Header cần tìm, ví dụ:
    # ^# (.+)$
    header_pattern = re.compile(
        rf"^{'#' * header_level}\s+(.+?)\s*$",
        re.MULTILINE
    )

    matches = list(header_pattern.finditer(content))

    if not matches:
        print("Không tìm thấy header.")
        return

    # Output folder
    if create_output_folder:
        output_dir = input_path.parent / f"{input_path.stem}-split"
        output_dir.mkdir(exist_ok=True)
    else:
        output_dir = input_path.parent

    num = init_num

    for i, match in enumerate(matches):
        header_title = match.group(1).strip()

        start = match.start()

        if i + 1 < len(matches):
            end = matches[i + 1].start()
        else:
            end = len(content)

        chunk = content[start:end].strip() + "\n"

        # Tạo filename
        if include_header_slug:
            slug = slugify(header_title)

            if slug:
                filename = f"{num:02d}-{slug}.md"
            else:
                filename = f"{num:02d}.md"
        else:
            filename = f"{num:02d}.md"

        output_path = output_dir / filename

        output_path.write_text(chunk, encoding="utf-8")

        print(f"{output_path.name}")

        num += 1


# =========================
# RUN
# =========================

split_markdown(input_file)
```

```python
#slit note

from pathlib import Path
import re


# =========================
# CONFIG
# =========================

input_file = Path("notes-vi.md")
output_dir = Path("note")


# =========================
# SLUG
# =========================

def slugify(text):
    """
    CHAPTER 1 -> chapter1
    The Life of Buddha -> the-life-of-buddha
    """

    text = text.strip().lower()

    # Bỏ markdown emphasis nếu có
    text = re.sub(r"[*_`]", "", text)

    # Giữ chữ, số và khoảng trắng
    text = re.sub(r"[^\w\s-]", "", text, flags=re.UNICODE)

    # Các header kiểu CHAPTER 1 -> chapter1
    text = re.sub(r"\b(chapter)\s+(\d+)\b", r"\1\2", text)

    # Khoảng trắng còn lại -> -
    text = re.sub(r"\s+", "-", text)

    # Nhiều dấu - -> một
    text = re.sub(r"-+", "-", text)

    return text.strip("-")


# =========================
# READ
# =========================

content = input_file.read_text(encoding="utf-8")

# Header level 2:
# ## CHAPTER 1
pattern = re.compile(r"(?m)^##[ \t]+(.+?)\s*$")

matches = list(pattern.finditer(content))

if not matches:
    raise ValueError("Không tìm thấy header level 2 (## ...)")


# =========================
# CREATE OUTPUT DIRECTORY
# =========================

output_dir.mkdir(parents=True, exist_ok=True)


# =========================
# SPLIT
# =========================

for i, match in enumerate(matches):

    title = match.group(1).strip()

    start = match.start()

    # Nội dung kết thúc trước header ## tiếp theo
    if i + 1 < len(matches):
        end = matches[i + 1].start()
    else:
        end = len(content)

    section = content[start:end].strip() + "\n"

    filename = slugify(title) + ".md"
    output_file = output_dir / filename

    output_file.write_text(section, encoding="utf-8")

    print(f"{output_file}")
```
